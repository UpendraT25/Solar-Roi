```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "../../datasets/electricity/cleaned_india_electricity.csv"
)

df.head()
```

### TOTAL CONSUMPTION TREND

```python
plt.figure(figsize=(15,6))

plt.plot(
    df["Dates"],
    df["total_consumption"]
)

plt.title("India Total Electricity Consumption")

plt.xlabel("Date")
plt.ylabel("Consumption")

plt.show()
```

### ROLLING AVERAGES

```python
plt.figure(figsize=(15,6))

plt.plot(
    df["Dates"],
    df["rolling_7_day_avg"],
    label="7 Day Avg"
)

plt.plot(
    df["Dates"],
    df["rolling_30_day_avg"],
    label="30 Day Avg"
)

plt.legend()

plt.show()
```
